```python
import pandas as pd

data = {
'Resume':[
'Python Machine Learning Data Analysis',
'Java Developer Web Development',
'Excel Communication Management'
],

'Category':[
'Data Science',
'Software Developer',
'Management'
]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(df['Resume'])

model = MultinomialNB()

model.fit(X, df['Category'])

print("Model Training Complete!")
```

```python
new_resume = ["Python Data Analysis Machine Learning"]

resume_data = vectorizer.transform(new_resume)

prediction = model.predict(resume_data)

print("Predicted Category =", prediction[0])
```
